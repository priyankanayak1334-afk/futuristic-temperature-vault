# Phase: Custom Data Structures - Futuristic Temperature Vault

### Mission
Build an efficient tracking system for a futuristic vault that stores temperature readings every second and instantly returns the minimum recorded temperature at any moment in **O(1) time**.

### Core Architecture
We use a **Dual-Stack Approach**:
1. **Primary Stack (`stack`)**: Stores every temperature reading in chronological order.
2. **Minimum Tracker Stack (`min_stack`)**: Keeps track of the minimum temperature at any given depth. 

This avoids scanning the collection during lookups, keeping our time complexity bound to constant time.


```python
import random
from typing import Optional

class TemperatureVaultStack:
    """
    A custom data structure for a futuristic temperature vault.
    Provides O(1) tracking for push, pop, top, and getMin operations.
    """
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, temp: float) -> None:
        """Pushes a new temperature reading into the vault."""
        self.stack.append(temp)
        # If min_stack is empty or the new temp is less than or equal to 
        # the current minimum, push it onto the min_stack.
        if not self.min_stack or temp <= self.min_stack[-1]:
            self.min_stack.append(temp)

    def pop(self) -> Optional[float]:
        """Removes and returns the most recent temperature reading."""
        if not self.stack:
            return None
        
        popped_val = self.stack.pop()
        # If the popped value matches the current minimum, remove it from min_stack too
        if popped_val == self.min_stack[-1]:
            self.min_stack.pop()
        return popped_val

    def top(self) -> Optional[float]:
        """Returns the most recent temperature reading without removing it."""
        return self.stack[-1] if self.stack else None

    def get_min(self) -> Optional[float]:
        """Instantly retrieves the minimum recorded temperature in the vault."""
        return self.min_stack[-1] if self.min_stack else None

print("🚀 TemperatureVaultStack class successfully defined!")
```

## Automated Testing with Random Streams

Now, we will simulate a streaming sensor network by pushing **10 random temperature values** (ranging from -50°C to 150°C) into our vault. 
We will verify that `get_min()` accurately adjusts dynamically with every ingestion and deletion step.


```python
# Initialize our vault instance
vault = TemperatureVaultStack()

print("--- Step 1: Ingesting Data Stream ---")
# Generate 10 random simulated temperature readings
simulated_temps = [round(random.uniform(-50.0, 150.0), 2) for _ in range(10)]
print(f"Generated Sequence: {simulated_temps}\n")

# Push each temperature and verify min lookup speed
for t in simulated_temps:
    vault.push(t)
    print(f"📥 Pushed: {t:6.2f}°C | ❄️  Current Min: {vault.get_min():6.2f}°C")
    
print("\n--- Step 2: Evicting Data (Pops) ---")
# Pop half of the data points to observe minimum rollback consistency
for i in range(5):
    popped = vault.pop()
    print(f"📤 Popped: {popped:6.2f}°C | 🔝 New Top: {vault.top():6.2f}°C | ❄️  Current Min: {vault.get_min():6.2f}°C")
```
